## Pushing to GitHub

To push your changes to a GitHub repository, you'll need to:

1.  **Configure Git credentials** with your `GITHUB_TOKEN`.
2.  **Add and commit** your changes.
3.  **Push** to your remote repository.

Make sure your `GITHUB_TOKEN` is available as an environment variable or stored securely (e.g., in Colab secrets if you're in Colab, or in your `.bashrc`/`.zshrc` or a `.env` file for local development). For this example, we assume `GITHUB_TOKEN` is an environment variable.

```python
import os
from google.colab import userdata

# Get the GitHub token from Colab's secret manager
github_token = userdata.get('GITHUB_TOKEN')

if not github_token:
    raise ValueError("GITHUB_TOKEN secret not found in Colab. Please add it to Colab secrets.")

# Configure Git to use the token for authentication
# Replace <YOUR_GITHUB_USERNAME> with your actual GitHub username
# Replace <YOUR_REPO_NAME> with the name of your repository

# Set Git email and username (optional but good practice)
!git config --global user.email "your_email@example.com" # Replace with your email
!git config --global user.name "Your Name" # Replace with your name

# This command sets up the credential helper for HTTPS, using the token
# for future operations.

# IMPORTANT: This assumes you have already initialized a Git repository
# and linked it to your remote GitHub repository (e.g., git remote add origin <repo_url>)

# If you are in a new Colab environment or haven't cloned a repo, you'll need to do the following:
# 1. !git init
# 2. !git remote add origin https://oauth2:$github_token@github.com/<YOUR_GITHUB_USERNAME>/<YOUR_REPO_NAME>.git
# 3. !git pull origin main (if cloning an existing repo)

# For now, we'll just set the global credential helper for the session
# This command sets the credential helper to use a token if found in the URL
# For simpler token handling, you can directly embed it in the remote URL as shown above (step 2)
print("Git configuration for GitHub token completed (credential helper set).")
print("Please ensure your Git repository is initialized and linked to your remote using the appropriate commands.")
```

### Add, Commit, and Push Changes

## Create a New Repository, Commit, and Push

Follow these steps to initialize a new Git repository in your current working directory (e.g., `/content/` in Colab), link it to a *new* or *empty* GitHub repository, and push your notebook.

**Before proceeding:**

1.  **Ensure `GITHUB_TOKEN` is set in Colab secrets.** (You've already addressed this in cell `d3e2163c`).
2.  **Decide on your GitHub repository name.** You'll use this in the steps below.
3.  **Make sure you are in the directory you want to initialize as a Git repository.** In Colab, this is typically `/content/`.

```python
# 1. Initialize a new Git repository
# This command creates a new .git subdirectory in your current directory.
!git init

print("Git repository initialized.")
```

```python
# 2. Add the remote origin to your GitHub repository

# IMPORTANT: Make sure you have run the 'd3e2163c' cell to get the github_token.
# If you restart the kernel, you might need to re-run 'd3e2163c' first.

# Replace <YOUR_GITHUB_USERNAME> with your GitHub username
# Replace <YOUR_REPO_NAME> with the name of your new or empty repository on GitHub.

# Assuming github_token is available from the execution of d3e2163c
github_token = userdata.get('GITHUB_TOKEN') # Re-get token just in case
!git remote add origin https://oauth2:{github_token}@github.com/EmmanuelM0147/Intro.git

print("Remote origin added successfully!")
```

```python
# Rename the local 'master' branch to 'main'
!git branch -M main

print("Local branch renamed from 'master' to 'main'.")
```

```python
# 3. Add your files to the staging area
# This adds all changes in the current directory to be committed.
!git add .

print("All changes added to staging.")
```

```python
# 4. Commit your changes
# This records the snapshot of your files in the repository history.
!git commit -m "Initial commit of Colab notebook" # Use a descriptive commit message

print("Changes committed.")
```

```python
# 5. Push your committed changes to GitHub
# The '-u' flag sets the upstream branch, so future 'git push' commands can be simpler.
# Replace 'main' with your target branch name if it's different (e.g., 'master').
!git push -u origin main

print("Changes pushed to GitHub successfully!")
```
